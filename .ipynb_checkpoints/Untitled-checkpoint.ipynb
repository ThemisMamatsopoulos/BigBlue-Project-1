```python
import pandas as pd
```

```python
df = pd.read_csv("archive.zip")
```

```python
df
```

```python
df.info()
```

```python
#Conver Order date to proper type and create 2 more columns for date and month
df['OrderDate']=pd.to_datetime(df['OrderDate'], errors='coerce')
df['year'] = df['OrderDate'].dt.year
df['month'] = df['OrderDate'].dt.month

df['OrderStatus'].unique() ### categories are --> 'Delivered', 'Cancelled', 'Pending', 'Shipped', 'Returned']
#Clean status - kept only delivered  got rid of 'Cancelled', 'Pending', 'Shipped', 'Returned' -- cleare around 25% of the dataset
df = df[df['OrderStatus'] == 'Delivered']
```

```python
len(df)
```

```python
Country_Categories = df.groupby(['Country','Category']).agg(
     sum_of_sales=('TotalAmount', 'sum'),
     Count_of_orders=('OrderID', 'count')
).sort_values(by='sum_of_sales',ascending=False).reset_index()
```

```python
Countries = df['Country'].unique()
Categories = df['Category'].unique()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sub_df = Country_Categories[Country_Categories['Country'] == 'United States']
```

```python
sub_df
```

```python
Categories
```

```python
fig, ax = plt.subplots()
ax.bar(sub_df['Category'], sub_df['sum_of_sales'], label='Categories')
ax.set_xticklabels(sub_df['Category'], rotation=90)
ax.set_ylim(7500000, 8200000)
ax1 = ax.twinx()
ax1.plot(sub_df['Category'], sub_df['Count_of_orders'], color='r')
ax.set_ylabel("Sales")
plt.show()
```
